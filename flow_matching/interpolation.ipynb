```python
import torch
import torch.nn as nn
from torchvision.datasets import CIFAR10
from torchvision.transforms import ToTensor
import matplotlib.pyplot as plt
```

```python
cifar = CIFAR10(
    '../data', download=True, transform=ToTensor()
)
```

```python
plt.imshow(cifar[3][0].permute(1, 2, 0))
```

```python
img = cifar[3][0]
```

```python
def interpolate(noise, target, time):
    interpolated = (target - noise) * time + noise
    return interpolated 
```

```python
gurt = interpolate(torch.rand((3, 32, 32)), img, 1.0)
plt.imshow(gurt.permute(1,2,0))
```
