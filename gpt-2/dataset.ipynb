```python
from datasets import load_dataset, load_dataset_builder
```

```python
builder = load_dataset_builder("openwebtext")
print(builder.info.splits["train"].num_examples)
```

```python
ds = load_dataset("openwebtext", split="train", streaming=True)
```

```python
docs = []
```

```python
iterator = iter(ds)
```

```python
for i in range(500_000):
    text = next(iterator)
    docs.append(text["text"])
    print(i, end="\r")
```

```python
unique_chars = set()
data_length = 0
```

```python
for i, doc in enumerate(docs):
    unique_chars.update(list(doc))
    data_length += len(doc)
    print(i, end="\r")
```

```python
import pickle
```

```python
with open("./vocab.pkl", 'wb') as f:
    pickle.dump(sorted(list(unique_chars)), f)
```

```python
token_to_index = {}
index_to_token = {}
```

```python
for i, word in enumerate(sorted(list(unique_chars))):
    token_to_index[word] = i
    index_to_token[i] = word
```

```python
with open("./vocab.pkl", 'wb') as f:
    pickle.dump({
        "tokens": sorted(list(unique_chars)),
        "token_to_index": token_to_index,
        "index_to_token": index_to_token
    }, f)
```

```python
sorted_unique_chars = sorted(list(unique_chars))
sorted_unique_chars.insert(0, "<eos>")
sorted_unique_chars.insert(0, "<sos>")
sorted_unique_chars.insert(0, "<pad>")
```

```python
for i, word in enumerate(sorted_unique_chars):
    token_to_index[word] = i
    index_to_token[i] = word
```

```python
with open("./vocab.pkl", 'wb') as f:
    pickle.dump({
        "tokens": sorted(list(unique_chars)),
        "token_to_index": token_to_index,
        "index_to_token": index_to_token
    }, f)
```

```python
tokenized_docs = []
```

```python
max_len = 0
```

```python
for doc in docs:
    # print(len(doc))
    if len(doc) > max_len:
        max_len = len(doc)
```

```python
max_len
```

```python
for i, doc in enumerate(docs):
    tokenized_doc = []
    tokenized_doc.append(0)
    for c in list(doc):
        tokenized_doc.append(token_to_index[c])

    tokenized_doc.append(1)

    tokenized_docs.append(tokenized_doc)

    print(i, end="\r")
```

```python
with open("data.pkl", 'wb') as f:
    pickle.dump(tokenized_docs, f)
```
